```python
import pandas as pd
```

```python
df = pd.read_csv('/home/devansh-roy/Desktop/New Folder/Data/phrasebank.csv',
                 header=None,
                 names=['sentiment', 'sentence'],
                 encoding='latin-1',
                 quotechar='"')
```

```python
print(df.shape)
print(df.head())
print(df['sentiment'].value_counts())
```

```python
df.describe()
```

```python
import re

df['sentence'] = df['sentence'].str.replace(r'[^a-zA-Z0-9\s]', '', regex=True)

df['sentence'] = df['sentence'].str.lower()

df['sentence'] = df['sentence'].str.strip()

df['label'] = df['sentiment'].map({'positive': 1, 'neutral': 0, 'negative': -1})

print(df.head(5))
print(df['label'].value_counts())
```
